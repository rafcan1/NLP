```python
#Libraries
import nltk
import string
import re
from collections import Counter
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
```

```python
#read the file
text = ""
with open('100-0.txt', 'r') as file:
  text = file.read()
```

1. Count the number of words in Shakespeare's works (1 point)
Tip: use word_tokenize
Expected Output:
Number of words: 1016743

```python
text_without_punctuation = text.translate(str.maketrans('', '', string.punctuation))
words = word_tokenize(text_without_punctuation)
#print(words)
word_count = len(words)
print(f'Number of words: {word_count}')
```

2. Print min, max, average and median number of characters in a word (1 point)
Expected Output:
Minimum word length: 1 characters
Maximum word length: 42 characters
Average word length: 4.04 characters
Median word length: 4 characters

```python
word_lengths = [len(word) for word in words]
min_length = min(word_lengths)
max_length = max(word_lengths)
average_length = sum(word_lengths) / len(word_lengths)
median_length = sorted(word_lengths)[len(word_lengths) // 2]
print(f'Minimum word length: {min_length} characters')
print(f'Maximum word length: {max_length} characters')
print(f'Average word length: {average_length:.2f} characters')
print(f'Median word length: {median_length} characters')
```

3. Count the number of unique lowercase stemmed words in Shakespeare's
works (1 point)
Tip: use PorterStemmer()
Expected Output:
Unique lowercase words: 16959

```python
cleaned_sentence = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
tokens = word_tokenize(cleaned_sentence.lower())
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(token) for token in tokens]
unique_stems = set(stemmed_words)
print(f'Unique lowercase words: {len(unique_stems)}')
```

4. Count the number of sentences and unique sentences in Shakespeare's
works (1 point)
Expected Output:
Numer of sentences: 104818 unique sentences: 70375


```python
sentences = sent_tokenize(text)
print(f'Numer of sentences: {len(sentences)} unique sentences: {len(set(sentences))}')
print(sentences)
```
