```python
"""
1. Tokenization (1 point)
Use spaCy to tokenize a given sample text into individual words and sentences.
Tip: use en_core_web_sm language model

Expected Output:
Word Tokenization:
[Hello, !, Welcome, to, the, world, of, Natural, Language, Processing, ., Let, 's,
learn, spaCy, together, .]
Sentence Tokenization:
[Hello!, Welcome to the world of Natural Language Processing., Let's learn
spaCy together.]
"""
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn spaCy together."
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
print("Word Tokenization:")
print([token.text for token in doc])
print("\nSentence Tokenization:")
print([sent.text for sent in doc.sents])
```

```python
"""
2. Removing Stopwords (1 point)
Filter out stopwords from a given text using spaCy.
Given
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn
spaCy together."
Expected Output:
Stopwords Removed: ['Hello', '!', 'Welcome', 'world', 'Natural', 'Language',
'Processing', '.', 'Let', 'learn', 'spaCy', '.']
"""
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn spaCy together."
doc = nlp(text)
stopwords_removed = [token.text for token in doc if not token.is_stop]
print("Stopwords Removed:", stopwords_removed)
```

```python
"""
3. Lemmatization (1 point)
Use spaCy to lemmatize each word in a sample text.
Given
words = ["running", "jumps", "easily", "programming", "programmers"]
Expected Output:
Lemmatized Words: ['run', 'jump', 'easily', 'programming', 'programmer']
"""
words = ["running", "jumps", "easily", "programming", "programmers"]
lemmatized_words = [token.lemma_ for token in nlp(" ".join(words))]
print("Lemmatized Words:", lemmatized_words)
```

```python
"""
4. Part-of-Speech Tagging (1 point)
Perform extended part-of-speech tagging (POS-tags) on a sample sentence
using spaCy.
Given
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn
spaCy together."
Expected Output:
Part-of-Speech Tagging:
[('Hello', 'INTJ', 'UH'), ('!', 'PUNCT', '.'), ('Welcome', 'NOUN', 'NN'), ('to', 'ADP', 'IN'),
('the', 'DET', 'DT'), ('world', 'NOUN', 'NN'), ('of', 'ADP', 'IN'), ('Natural', 'PROPN',
'NNP'), ('Language', 'PROPN', 'NNP'), ('Processing', 'PROPN', 'NNP'), ('.', 'PUNCT',
'.'), ('Let', 'VERB', 'VB'), ("'s", 'PRON', 'PRP'), ('learn', 'VERB', 'VB'), ('spaCy', 'ADJ',
'JJ'), ('together', 'ADV', 'RB'), ('.', 'PUNCT', '.')]
"""
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn spaCy together."
doc = nlp(text)
pos_tags = [(token.text, token.pos_, token.tag_) for token in doc]
print("Part-of-Speech Tagging:")
print(pos_tags)
```

```python
"""
5. Named Entity Recognition (1 point)
Identify named entities in a sample text using spaCy's named entity recognition
(NER).
Given
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn
spaCy together."
Expected Output:
Named Entity Recognition: [('Natural Language Processing', 'WORK_OF_ART')]
"""
text = "Hello! Welcome to the world of Natural Language Processing. Let's learn spaCy together."
doc = nlp(text)
named_entities = [(ent.text, ent.label_) for ent in doc.ents]
print("Named Entity Recognition:")
print(named_entities)
```

```python
"""
6. Sentence Similarity (1 point)
Use spaCy to compare the similarity between two sentences.
Given
sentence1 = "I love apples."
sentence2 = "I enjoy eating fruit."
Expected Output:
Similarity between the sentences: 0.71
"""
sentence1 = "I love apples."
sentence2 = "I enjoy eating fruit."
doc1 = nlp(sentence1)
doc2 = nlp(sentence2)
similarity = doc1.similarity(doc2)
print("Similarity between the sentences:", similarity)
```
