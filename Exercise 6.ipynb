1. Count the number of words in Shakespeare's works (1 point)
Tip: use token.is_alpha and token.text
Expected Output:
Number of words: 178201

```python
def process_in_chunks(text, chunk_size=100000):
    docs = []
    for i in range(0, len(text), chunk_size):
        chunk = text[i:i+chunk_size]
        doc = nlp(chunk)
        docs.append(doc)
    return docs
chunky_text = process_in_chunks(text)
```

```python

alpha_tokens = [token.text for token in chunky_text if token.is_alpha]
```
