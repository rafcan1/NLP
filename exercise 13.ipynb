```python
import wikipedia
import spacy
```

```python
def search_wikipedia(query):
    """Search Wikipedia and return the summary of the top result."""
    try:
        results = wikipedia.search(query)
        if results:
            summary = wikipedia.summary(results[0], sentences=15)
            return summary
            print(summary)
        else:
            return "No results found."
    except wikipedia.exceptions.DisambiguationError as e:
        return f"Disambiguation error: {e.options}"
    except Exception as e:
        return f"An error occurred: {e}"
```

```python
def answer_the_question(question):
  nlp = spacy.load("en_core_web_sm")

  def extract_noun_phrases(question):
      """Extract noun phrases from a question."""
      doc = nlp(question)
      return [chunk.text for chunk in doc.noun_chunks]

  noun_phrases = extract_noun_phrases(question)
  # print("Noun Phrases:", noun_phrases)

  # for token in nlp(question):
  #   print(token, token.dep_, token.tag_)

  dependencies = ["pobj", "dobj", "nsubj", "attr", "root"]
  pobj = None
  nsubj = None

  for token in nlp(question):
    if token.dep_ in dependencies:
        if token.dep_ == "nsubj":
            nsubj = token.lemma_
        else:
            pobj = token.text

  if not pobj:
    pobj = nsubj
  result = search_wikipedia(pobj)
  print(result)

  doc = nlp(result)
  answer_sentecne = None
  for sentence in doc.sents:
    for token in sentence:
      if nsubj == token.lemma_:
        answer_sentecne = sentence

  if answer_sentecne:
    print(answer_sentecne)
    doc = nlp(answer_sentecne.text)
    for token in doc:
      if token.tag_ == "NNP" and token.text.lower() not in [pobj.lower(), nsubj.lower()]:
        print(f"{token.text}")
```

```python
question= input("Enter your question: ")
answer_the_question(question)
```
